```python
# Sentence Embeddings using BERT and GPT-2

!pip -q install transformers torch

import torch
from transformers import AutoTokenizer, AutoModel

# Sample sentences
sentences = [
    "Natural language processing is interesting.",
    "Machine learning is a part of artificial intelligence."
]

# Load BERT model
bert_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
bert_model = AutoModel.from_pretrained("bert-base-uncased")

# Load GPT-2 model
gpt_tokenizer = AutoTokenizer.from_pretrained("gpt2")
gpt_model = AutoModel.from_pretrained("gpt2")

# Set padding token for GPT-2
gpt_tokenizer.pad_token = gpt_tokenizer.eos_token
gpt_model.config.pad_token_id = gpt_tokenizer.eos_token_id

# Generate BERT sentence embeddings
def get_bert_embeddings(texts):
    inputs = bert_tokenizer(
        texts, return_tensors="pt",
        padding=True, truncation=True
    )
    with torch.no_grad():
        outputs = bert_model(**inputs)
    # Mean pooling over non-padding tokens
    mask = inputs["attention_mask"].unsqueeze(-1)
    embeddings = (outputs.last_hidden_state * mask).sum(1) / mask.sum(1)
    return embeddings

# Generate GPT-2 sentence embeddings
def get_gpt2_embeddings(texts):
    inputs = gpt_tokenizer(
        texts, return_tensors="pt",
        padding=True, truncation=True
    )
    with torch.no_grad():
        outputs = gpt_model(**inputs)
    # Mean pooling over non-padding tokens
    mask = inputs["attention_mask"].unsqueeze(-1)
    embeddings = (outputs.last_hidden_state * mask).sum(1) / mask.sum(1)
    return embeddings

# Calculate embeddings
bert_embeddings = get_bert_embeddings(sentences)
gpt2_embeddings = get_gpt2_embeddings(sentences)

# Display results
for i, sentence in enumerate(sentences):
    print("\nSentence:", sentence)
    print("BERT Embedding Shape:", bert_embeddings[i].shape)
    print("BERT Vector (first 10 values):",
          bert_embeddings[i][:10].numpy().round(3))
    print("GPT-2 Embedding Shape:", gpt2_embeddings[i].shape)
    print("GPT-2 Vector (first 10 values):",
          gpt2_embeddings[i][:10].numpy().round(3))
```
