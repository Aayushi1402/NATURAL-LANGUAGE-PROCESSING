```python
# Opinion Mining on Product/Service Reviews Dataset

from textblob import TextBlob
import pandas as pd

# Sample product reviews dataset
reviews = [
    "The product is excellent and very useful",
    "The service was very poor",
    "I am happy with the quality",
    "The product is average",
    "Customer support was amazing",
    "The product is disappointing"
]

# Perform opinion mining
results = []

for review in reviews:
    polarity = TextBlob(review).sentiment.polarity

    if polarity > 0:
        sentiment = "Positive"
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    results.append({
        "Review": review,
        "Polarity": round(polarity, 2),
        "Opinion": sentiment
    })

# Display results
df = pd.DataFrame(results)

print("Opinion Mining Results:\n")
print(df.to_string(index=False))

print("\nOverall Sentiment Summary:")
print(df["Opinion"].value_counts())
```
