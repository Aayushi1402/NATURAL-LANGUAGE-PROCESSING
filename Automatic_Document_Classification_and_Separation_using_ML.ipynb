```python
# Automatic Document Classification and Separation using ML

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

# Sample documents and categories
documents = [
    "The team won the football match",
    "The player scored a goal",
    "Artificial intelligence improves technology",
    "Machine learning uses data",
    "The stock market prices increased",
    "Banks provide financial services"
]

labels = [
    "Sports", "Sports",
    "Technology", "Technology",
    "Finance", "Finance"
]

# Create and train the classification model
model = make_pipeline(
    TfidfVectorizer(),
    MultinomialNB()
)
model.fit(documents, labels)

# New documents to classify
new_documents = [
    "The team played a great match",
    "AI models learn from data",
    "The bank offers investment services"
]

# Classify and separate documents
separated_documents = {}

for document in new_documents:
    category = model.predict([document])[0]
    separated_documents.setdefault(category, []).append(document)

# Display classified documents
print("Document Classification and Separation:\n")

for category, docs in separated_documents.items():
    print(f"Category: {category}")
    for doc in docs:
        print("-", doc)
    print()
```
