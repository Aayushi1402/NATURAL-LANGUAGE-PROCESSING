```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import sent_tokenize, word_tokenize

text = "AI is transforming the world. NLP helps computers understand language."

print("Sentence Tokenization using NLTK:")
print(sent_tokenize(text))

print("\nWord Tokenization using NLTK:")
print(word_tokenize(text))
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "AI is transforming the world. NLP helps computers understand language."

doc = nlp(text)

print("Sentence Tokenization using spaCy:")
for sentence in doc.sents:
    print(sentence.text)

print("\nWord Tokenization using spaCy:")
for token in doc:
    print(token.text)
```
