```python
# Information Extraction from Structured and Unstructured Documents

import re
import pandas as pd
import spacy

# Install and load spaCy model
!python -m spacy download en_core_web_sm -q

nlp = spacy.load("en_core_web_sm")

# 1. Structured document (CSV-like data)
data = {
    "Name": ["Rahul", "Priya", "Amit"],
    "Age": [21, 22, 20],
    "City": ["Delhi", "Mumbai", "Pune"]
}

df = pd.DataFrame(data)

print("STRUCTURED DATA:")
print(df)

# 2. Unstructured document
text = """
Rahul Sharma works at Google in Delhi.
He joined the company in 2022.
Contact email: rahul@example.com
"""

print("\nUNSTRUCTURED TEXT:")
print(text)

# Extract named entities using spaCy
doc = nlp(text)

print("\nEXTRACTED NAMED ENTITIES:")
for ent in doc.ents:
    print(ent.text, "->", ent.label_)

# Extract email addresses using Regex
emails = re.findall(
    r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b',
    text
)

print("\nEXTRACTED EMAILS:")
print(emails)

# Extract years using Regex
years = re.findall(r'\b(?:19|20)\d{2}\b', text)

print("\nEXTRACTED YEARS:")
print(years)
```
