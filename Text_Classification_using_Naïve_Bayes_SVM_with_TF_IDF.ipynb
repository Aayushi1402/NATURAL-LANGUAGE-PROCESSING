```python
# Text Classification using Naïve Bayes and SVM with TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

# Sample dataset
texts = [
    "I love this movie",
    "This film is amazing",
    "The movie was excellent",
    "I enjoyed the film",
    "This movie is fantastic",
    "I hate this movie",
    "This film is terrible",
    "The movie was boring",
    "I disliked the film",
    "This movie is awful"
]

labels = [
    "Positive", "Positive", "Positive", "Positive", "Positive",
    "Negative", "Negative", "Negative", "Negative", "Negative"
]

# Naïve Bayes model with TF-IDF
nb_model = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('classifier', MultinomialNB())
])

# SVM model with TF-IDF
svm_model = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('classifier', LinearSVC(random_state=42))
])

# Train both models
nb_model.fit(texts, labels)
svm_model.fit(texts, labels)

# Test sentences
test_texts = [
    "This movie is wonderful",
    "The film was boring"
]

# Predictions using Naïve Bayes
nb_predictions = nb_model.predict(test_texts)

# Predictions using SVM
svm_predictions = svm_model.predict(test_texts)

# Display results
print("Naïve Bayes Classification:")
for text, prediction in zip(test_texts, nb_predictions):
    print(text, "->", prediction)

print("\nSVM Classification:")
for text, prediction in zip(test_texts, svm_predictions):
    print(text, "->", prediction)

# Training accuracy (demonstration only)
print("\nNaïve Bayes Training Accuracy:",
      round(accuracy_score(labels, nb_model.predict(texts)) * 100, 2), "%")

print("SVM Training Accuracy:",
      round(accuracy_score(labels, svm_model.predict(texts)) * 100, 2), "%")
```
