```python
# Implementing Attention Mechanism (Basic Custom Model)

import tensorflow as tf
import numpy as np
from tensorflow.keras.layers import Input, Dense, Softmax
from tensorflow.keras.models import Model

# Sample input data: 3 samples, 4 time steps, 5 features
X = np.random.rand(3, 4, 5).astype(np.float32)

# Input layer
inputs = Input(shape=(4, 5))

# Calculate attention scores
scores = Dense(1, activation='tanh')(inputs)

# Normalize scores using Softmax
attention_weights = Softmax(axis=1, name="attention_weights")(scores)

# Apply attention weights to input
weighted_input = tf.keras.layers.Multiply()([inputs, attention_weights])

# Sum weighted inputs across time steps
context_vector = tf.keras.layers.Lambda(
    lambda x: tf.reduce_sum(x, axis=1)
)(weighted_input)

# Output layer
output = Dense(1, activation='sigmoid')(context_vector)

# Build model
model = Model(inputs, output)

# Compile model
model.compile(optimizer='adam', loss='binary_crossentropy',
              metrics=['accuracy'])

# Display model summary
model.summary()

# Sample labels
y = np.array([0, 1, 0])

# Train model
model.fit(X, y, epochs=5, verbose=1)

# Predict
prediction = model.predict(X, verbose=0)

print("\nPredictions:")
print(prediction.round(3))

# Display attention weights
attention_model = Model(inputs, model.get_layer("attention_weights").output)
weights = attention_model.predict(X, verbose=0)

print("\nAttention Weights:")
print(weights.squeeze(-1).round(3))
```
