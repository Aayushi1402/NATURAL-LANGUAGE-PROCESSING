```python
# Topic Modeling using Latent Dirichlet Allocation (LDA)

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

# Sample documents
documents = [
    "Machine learning and artificial intelligence are growing rapidly",
    "Deep learning uses neural networks for image recognition",
    "Football and cricket are popular sports",
    "Players practice football and cricket every day",
    "Artificial intelligence uses machine learning algorithms",
    "Sports teams train players for football matches"
]

# Convert text into a document-term matrix
vectorizer = CountVectorizer(stop_words='english')
dtm = vectorizer.fit_transform(documents)

# Create and train the LDA model
lda = LatentDirichletAllocation(
    n_components=2,
    random_state=42
)
lda.fit(dtm)

# Display topics
words = vectorizer.get_feature_names_out()

print("Topics discovered by LDA:\n")
for topic_idx, topic in enumerate(lda.components_):
    top_words = [words[i] for i in topic.argsort()[-5:][::-1]]
    print(f"Topic {topic_idx + 1}: {', '.join(top_words)}")

# Display topic distribution for each document
print("\nTopic Distribution for Documents:")
distributions = lda.transform(dtm)

for i, distribution in enumerate(distributions):
    print(f"Document {i + 1}: {distribution.round(3)}")
```
