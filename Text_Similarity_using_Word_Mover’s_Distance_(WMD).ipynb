```python
# Text Similarity using Word Mover's Distance (WMD)

# We install POT (Python Optimal Transport) which is more reliable and faster to install than pyemd
!pip -q install gensim POT numpy

import nltk
import gensim.downloader as api
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Load pretrained Word2Vec embeddings
print("Loading word embeddings...")
model = api.load("word2vec-google-news-300")

# Sample documents
text1 = "I love machine learning and artificial intelligence"
text2 = "I enjoy deep learning and AI technology"

# Preprocess text
stop_words = set(stopwords.words('english'))

def preprocess(text):
    words = word_tokenize(text.lower())
    return [word for word in words
            if word.isalpha() and word not in stop_words
            and word in model.key_to_index]

words1 = preprocess(text1)
words2 = preprocess(text2)

# Calculate Word Mover's Distance using gensim's built-in wmdistance (which uses POT internally if pyemd is missing)
distance = model.wmdistance(words1, words2)

# Display results
print("\nText 1:", text1)
print("Text 2:", text2)
print("\nWord Mover's Distance:", round(distance, 4))

if distance != float('inf'):
    print("Interpretation: Lower distance means greater semantic similarity.")
else:
    print("Distance could not be calculated for these documents.")
```
