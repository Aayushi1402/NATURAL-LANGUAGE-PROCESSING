```python
# N-Gram Model (Unigram, Bigram, Trigram) Generation

from sklearn.feature_extraction.text import CountVectorizer

# Sample corpus
corpus = [
    "Natural language processing is interesting",
    "Natural language processing is useful"
]

# Generate Unigrams
unigram = CountVectorizer(ngram_range=(1, 1))
unigram_matrix = unigram.fit_transform(corpus)

print("Unigrams:")
print(unigram.get_feature_names_out())

# Generate Bigrams
bigram = CountVectorizer(ngram_range=(2, 2))
bigram_matrix = bigram.fit_transform(corpus)

print("\nBigrams:")
print(bigram.get_feature_names_out())

# Generate Trigrams
trigram = CountVectorizer(ngram_range=(3, 3))
trigram_matrix = trigram.fit_transform(corpus)

print("\nTrigrams:")
print(trigram.get_feature_names_out())
```
