```python
# Machine Translation using MarianMT (Hugging Face)

!pip -q install transformers sentencepiece

from transformers import MarianMTModel, MarianTokenizer

# Select translation model: English to French
model_name = "Helsinki-NLP/opus-mt-en-fr"

# Load tokenizer and model
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

# Sample English sentences
sentences = [
    "Hello, how are you?",
    "I love learning artificial intelligence.",
    "Natural language processing is very useful."
]

# Translate sentences
inputs = tokenizer(sentences, return_tensors="pt",
                   padding=True, truncation=True)

translated = model.generate(**inputs)

# Decode and display translations
translations = tokenizer.batch_decode(
    translated, skip_special_tokens=True
)

print("Machine Translation Results:\n")
for source, target in zip(sentences, translations):
    print("English:", source)
    print("French:", target)
    print()
```
