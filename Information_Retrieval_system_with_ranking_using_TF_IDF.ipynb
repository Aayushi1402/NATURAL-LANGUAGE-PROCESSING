```python
# Information Retrieval System with Ranking using TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sample document collection
documents = [
    "Machine learning is a branch of artificial intelligence",
    "Deep learning uses neural networks",
    "Natural language processing works with human language",
    "Artificial intelligence includes machine learning and deep learning",
    "Information retrieval helps users find relevant documents"
]

# User search query
query = "machine learning artificial intelligence"

# Convert documents and query into TF-IDF vectors
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)
query_vector = vectorizer.transform([query])

# Calculate similarity scores
scores = cosine_similarity(query_vector, tfidf_matrix).flatten()

# Rank documents by relevance
ranked_indices = scores.argsort()[::-1]

# Display ranked results
print("Search Query:", query)
print("\nRanked Documents:\n")

for rank, index in enumerate(ranked_indices, start=1):
    if scores[index] > 0:
        print(f"Rank {rank} | Score: {scores[index]:.3f}")
        print("Document:", documents[index])
        print()
```
