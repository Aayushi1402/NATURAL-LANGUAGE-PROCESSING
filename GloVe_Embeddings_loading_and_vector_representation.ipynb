```python
# GloVe Embeddings Loading and Vector Representation

import gensim.downloader as api

# Load pretrained GloVe embeddings
print("Loading GloVe model...")
model = api.load("glove-wiki-gigaword-50")

# Display vector representation of a word
word = "computer"

print("\nWord:", word)
print("\nGloVe Vector:")
print(model[word])

# Find similar words
print("\nMost Similar Words:")
for similar_word, score in model.most_similar(word, topn=5):
    print(similar_word, ":", round(score, 3))

# Calculate similarity between two words
word1 = "king"
word2 = "queen"

similarity = model.similarity(word1, word2)
print(f"\nSimilarity between {word1} and {word2}: {similarity:.3f}")
```
