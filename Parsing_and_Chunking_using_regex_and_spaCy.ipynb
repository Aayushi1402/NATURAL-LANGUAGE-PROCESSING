```python
# Parsing and Chunking using Regex and spaCy

import nltk
import spacy
import re
from nltk import word_tokenize, pos_tag, RegexpParser

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

# Load spaCy English model
!python -m spacy download en_core_web_sm

nlp = spacy.load("en_core_web_sm")

# Sample sentence
sentence = "The young boy is playing football in the park."

print("Original Sentence:")
print(sentence)

# 1. Parsing and Chunking using Regex (NLTK)
words = word_tokenize(sentence)
tagged_words = pos_tag(words)

# Define a regex grammar for noun phrases
grammar = r"NP: {<DT>?<JJ.*>*<NN.*>+}"

chunk_parser = RegexpParser(grammar)
chunk_tree = chunk_parser.parse(tagged_words)

print("\nRegex Chunking Result:")
print(chunk_tree)

# 2. Chunking using spaCy
doc = nlp(sentence)

print("\nspaCy Parsing and Chunking:")
for chunk in doc.noun_chunks:
    print("Noun Phrase:", chunk.text)

print("\nDependency Parsing:")
for token in doc:
    print(token.text, "->", token.dep_, "->", token.head.text)
```
