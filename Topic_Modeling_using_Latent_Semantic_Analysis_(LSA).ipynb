```python
# Topic Modeling using Latent Semantic Analysis (LSA)

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

# Sample documents
documents = [
    "Machine learning and artificial intelligence are useful",
    "Deep learning uses neural networks",
    "Football and cricket are popular sports",
    "Players practice football and cricket",
    "Artificial intelligence uses machine learning",
    "Sports teams train players for matches"
]

# Convert text into TF-IDF matrix
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(documents)

# Apply LSA using Truncated SVD
lsa = TruncatedSVD(n_components=2, random_state=42)
lsa_matrix = lsa.fit_transform(tfidf_matrix)

# Display topics
terms = vectorizer.get_feature_names_out()

print("Topics Discovered by LSA:\n")
for topic_idx, topic in enumerate(lsa.components_):
    top_words = [terms[i] for i in topic.argsort()[-5:][::-1]]
    print(f"Topic {topic_idx + 1}: {', '.join(top_words)}")

# Display document-topic representation
print("\nDocument-Topic Matrix:")
print(lsa_matrix.round(3))
```
